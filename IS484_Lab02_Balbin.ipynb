## 4.2. Representación de un problema mediante estados

**Parte A: Ficha del problema del repartidor**

| Elemento | Descripción del Problema |
| :--- | :--- |
| **Estado inicial** | El repartidor se encuentra en el **Almacén**. |
| **Estado objetivo** | El repartidor debe llegar a la **Zona Norte**. |
| **Acciones posibles** | Desplazarse hacia nodos adyacentes permitidos por el mapa (ej. ir de Almacén a Centro, o de Almacén a Terminal). |
| **Costo de cada acción** | 1 unidad de distancia/tramo recorrido entre dos puntos conectados directamente. |

**Parte B: Representar el mapa de la ciudad como un grafo**

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}

print("Mapa cargado en memoria:")
for nodo, vecinos in mapa.items():
    print(f" - {nodo} se conecta con: {vecinos}")
```

## 4.3 Implementación de BFS

**Búsqueda en Amplitud (Breadth-First Search)**
: El algoritmo BFS explora el grafo nivel por nivel. Utiliza una estructura de tipo cola (FIFO: el primero en entrar es el primero en salir), lo que garantiza encontrar el camino con menos pasos hacia el nodo objetivo.

**Traza paso a paso**

| Turno | Camino que se saca de la cola | Nodo actual | ¿Es la meta? | Se agrega a la cola |
| :--- | :--- | :--- | :--- | :--- |
| 1 | Almacén | Almacén | No | Almacén-Centro, Almacén-Terminal |
| 2 | Almacén-Centro | Centro | No | Almacén-Centro-Almacén, Almacén-Centro-ZonaNorte, Almacén-Centro-Mercado |
| 3 | Almacén-Terminal | Terminal | No | Almacén-Terminal-Almacén, Almacén-Terminal-Mercado |
| 4 | Almacén-Centro-Almacén | Almacén | No (ya visitado) | *(se descarta, no se expande)* |
| 5 | Almacén-Centro-ZonaNorte | ZonaNorte | Sí – fin | – |

 Nota de análisis: Fíjense en el Turno 1: cuando sacamos Almacén de la cola y vemos sus vecinos (Centro y Terminal), los agregamos al FINAL de la cola – no los exploramos todavía. Por eso en el Turno 2 procesamos Centro completo, y recién en el Turno 3 volvemos a Terminal. Eso es explorar por anillos.

```python
def bfs(grafo, inicio, objetivo):
    cola = [[inicio]]
    visitados = set()
    
    while cola:
        camino = cola.pop(0)
        nodo_actual = camino[-1]
        
        if nodo_actual == objetivo:
            return camino
            
        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            
            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]
                cola.append(nuevo_camino)
                
    return None

resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```
