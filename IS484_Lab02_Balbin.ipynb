## 4.2. Representación de un problema mediante estados

**Parte A: Ficha del problema del repartidor**

| Elemento | Descripción del Problema |
| :--- | :--- |
| **Estado inicial** | El repartidor se encuentra en el **Almacén**. |
| **Estado objetivo** | El repartidor debe llegar a la **Zona Norte**. |
| **Acciones posibles** | Desplazarse hacia nodos adyacentes permitidos por el mapa (ej. ir de Almacén a Centro, o de Almacén a Terminal). |
| **Costo de cada acción** | 1 unidad de distancia/tramo recorrido entre dos puntos conectados directamente. |

**Parte B: Representar el mapa de la ciudad como un grafo**

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}

print("Mapa cargado en memoria:")
for nodo, vecinos in mapa.items():
    print(f" - {nodo} se conecta con: {vecinos}")
```

## 4.3 Implementación de BFS

**Búsqueda en Amplitud (Breadth-First Search)**
: El algoritmo BFS explora el grafo nivel por nivel. Utiliza una estructura de tipo cola (FIFO: el primero en entrar es el primero en salir), lo que garantiza encontrar el camino con menos pasos hacia el nodo objetivo.

**Traza paso a paso**

| Turno | Camino que se saca de la cola | Nodo actual | ¿Es la meta? | Se agrega a la cola |
| :--- | :--- | :--- | :--- | :--- |
| 1 | Almacén | Almacén | No | Almacén-Centro, Almacén-Terminal |
| 2 | Almacén-Centro | Centro | No | Almacén-Centro-Almacén, Almacén-Centro-ZonaNorte, Almacén-Centro-Mercado |
| 3 | Almacén-Terminal | Terminal | No | Almacén-Terminal-Almacén, Almacén-Terminal-Mercado |
| 4 | Almacén-Centro-Almacén | Almacén | No (ya visitado) | *(se descarta, no se expande)* |
| 5 | Almacén-Centro-ZonaNorte | ZonaNorte | Sí – fin | – |

 Nota de análisis: Fíjense en el Turno 1: cuando sacamos Almacén de la cola y vemos sus vecinos (Centro y Terminal), los agregamos al FINAL de la cola – no los exploramos todavía. Por eso en el Turno 2 procesamos Centro completo, y recién en el Turno 3 volvemos a Terminal. Eso es explorar por anillos.

```python
def bfs(grafo, inicio, objetivo):
    cola = [[inicio]]
    visitados = set()
    
    while cola:
        camino = cola.pop(0)
        nodo_actual = camino[-1]
        
        if nodo_actual == objetivo:
            return camino
            
        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            
            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]
                cola.append(nuevo_camino)
                
    return None

resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```

## 4.4 Vectores y matrices con NumPy

**Fundamento Matemático y Representación Estructurada**

En los sistemas de Inteligencia Artificial y Aprendizaje Automático (Machine Learning), los conjuntos de datos tabulares se representan formalmente mediante estructuras algebraicas de arreglos multidimensionales (`ndarray` de NumPy):

* **Vector de características ($\vec{x}$):** Representa una única observación del entorno[cite: 6]. Cada posición o índice corresponde a una variable medida (por ejemplo: $[x_1 = \text{temperatura}, x_2 = \text{humedad}]$).
* **Matriz de datos ($X$):** Consiste en el apilamiento ordenado de $m$ vectores de observación de tamaño $n$. La dimensión resultante es $m \times n$, donde:
  * **Filas ($i$):** Muestras o registros individuales observados por los sensores.
  * **Columnas ($j$):** Atributos, dimensiones o características cuantitativas registradas.

---

 **Parte A: Estructura de Datos Tabular del Dataset**

| Índice de Observación | Sensor / Fila | Temperatura (°C) [$j=0$] | Humedad (%) [$j=1$] |
| :---: | :--- | :---: | :---: |
| 0 | Sensor 1 (`obs1`) | 22.5 | 60.0 |
| 1 | Sensor 2 (`obs2`) | 19.0 | 45.0 |
| 2 | Sensor 3 (`obs3`) | 25.3 | 70.0 |
| 3 | Sensor 4 (`obs4`) | 21.0 | 50.0 |

```python
import numpy as np

# Definición de observaciones individuales como vectores 1D (temperatura, humedad)
obs1 = np.array([22.5, 60.0])
obs2 = np.array([19.0, 45.0])
obs3 = np.array([25.3, 70.0])
obs4 = np.array([21.0, 50.0])

# Consolidación de observaciones en una matriz 2D (dataset)
dataset = np.array([obs1, obs2, obs3, obs4])

print("Matriz consolidada (dataset):\n", dataset)
print("\nDimensiones de la matriz (filas, columnas):", dataset.shape)
print("Tipo de dato interno:", dataset.dtype)
```

**Parte B: Indexación y Segmentación (Slicing) en Arreglos Bidimensionales**

Para extraer elementos específicos de una matriz bidimensional con NumPy, se utiliza la notación de corchetes `dataset[filas, columnas]`:

1. **Selección de fila completa (`dataset[i]` o `dataset[i, :]`):** 
   Extrae el vector $1\text{D}$ que describe íntegramente la observación seleccionada $i$[cite: 6].
2. **Selección de columna completa (`dataset[:, j]`):** 
   El operador de dos puntos `:` indica la inclusión de todas las filas, aislando únicamente la dimensión o variable $j$ a lo largo de todo el conjunto de datos[cite: 6].

```python
# Acceso a una fila completa: Primera observación (Sensor 1)
primera_obs = dataset[0]
print("Primera observación (fila 0):", primera_obs)

# Acceso a una columna completa: Atributo Temperatura de todas las observaciones
todas_temperaturas = dataset[:, 0]
print("Todas las temperaturas (columna 0):", todas_temperaturas)

# Acceso a una columna completa: Atributo Humedad de todas las observaciones
todas_humedades = dataset[:, 1]
print("Todas las humedades (columna 1):", todas_humedades)
```

**Interpretación Técnica de los Resultados**

| Operación | Sintaxis NumPy | Resultado Obtenido | Significado en el Espacio de Características |
| :--- | :--- | :--- | :--- |
| **Dimensión total** | `dataset.shape` | `(4, 2)` | El dataset contiene 4 observaciones distribuidas en 2 dimensiones cuantitativas[cite: 6]. |
| **Aislamiento de muestra** | `dataset[0]` | `[22.5, 60. ]` | Recupera el vector de estado del primer sensor: 22.5 °C y 60.0% de humedad[cite: 6]. |
| **Aislamiento de variable** | `dataset[:, 0]` | `[22.5, 19. , 25.3, 21. ]` | Vector que proyecta únicamente la distribución térmica de todas las muestras[cite: 6]. |
| **Aislamiento de variable** | `dataset[:, 1]` | `[60. , 45. , 70. , 50. ]` | Vector que proyecta únicamente la distribución de humedad de todas las muestras. |
